### Lab 6.1: Simple Calculator Functions
**Objective:** Define and use functions to perform basic arithmetic operations, practicing parameter passing and return values.

**Tasks:**
1. Define four functions: `add(a, b)`, `subtract(a, b)`, `multiply(a, b)`, `divide(a, b)`
2. Each function takes two numbers as parameters and returns the result
3. Add a docstring to each function
4. Menu-driven program with `while` loop
5. Handle division by zero
6. **Challenge:** `power(base, exponent)` with default parameter `exponent=2`

```python
def add(a, b):
    """Adds two numbers and returns the result."""
    return a + b

def subtract(a, b):
    """Subtracts second number from first and returns the result."""
    return a - b

def multiply(a, b):
    """Multiplies two numbers and returns the result."""
    return a * b

def divide(a, b):
    """Divides first number by second. Returns error message if division by zero."""
    if b == 0:
        return "Error: Division by zero"
    return a / b

def power(base, exponent=2):
    """Raises base to the power of exponent. Default exponent is 2 (squaring)."""
    return base ** exponent


while True:
    print("\n--- Simple Calculator ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Power (Challenge)")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ")

    if choice == '6':
        print("Exiting calculator. Goodbye!")
        break

    if choice == '1':
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            print(f"Result: {add(num1, num2)}")
        except ValueError:
            print("Invalid input. Please enter numbers.")

    elif choice == '2':
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            print(f"Result: {subtract(num1, num2)}")
        except ValueError:
            print("Invalid input. Please enter numbers.")

    elif choice == '3':
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            print(f"Result: {multiply(num1, num2)}")
        except ValueError:
            print("Invalid input. Please enter numbers.")

    elif choice == '4':
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            print(f"Result: {divide(num1, num2)}")
        except ValueError:
            print("Invalid input. Please enter numbers.")

    elif choice == '5':
        try:
            num1 = float(input("Enter base number: "))
            exp_input = input("Enter exponent (press Enter for default 2): ")
            if exp_input.strip() == "":
                print(f"Result: {power(num1)}")
            else:
                num2 = float(exp_input)
                print(f"Result: {power(num1, num2)}")
        except ValueError:
            print("Invalid input. Please enter numbers.")

    else:
        print("Invalid choice. Please select from 1-6.")
```

### Lab 6.2: Custom Module and Standard Library Usage
**Objective:** Create a custom Python module and import it, along with using standard library modules (`math`, `random`).

**Part 1: Create a Custom Module (`my_utils.py`)**

Using `%%writefile` magic command to create the module file.

```python
%%writefile my_utils.py
# my_utils.py (Custom Module)

def greet(name):
    """Prints a greeting message to the given name."""
    print(f"Hello, {name}!")

def is_prime(number):
    """Returns True if the number is prime, False otherwise."""
    if number <= 1:
        return False
    if number <= 3:
        return True
    if number % 2 == 0 or number % 3 == 0:
        return False
    i = 5
    while i * i <= number:
        if number % i == 0 or number % (i + 2) == 0:
            return False
        i += 6
    return True
```

**Part 2: Use the Custom Module and Standard Modules**

```python
import my_utils
import math
import random

# === ทดสอบ Custom Module ===
print("--- Testing Custom Module (my_utils) ---")
my_utils.greet("Alice")

test_numbers = [7, 10, 13, 20, 23]
for num in test_numbers:
    print(f"Is {num} prime? {my_utils.is_prime(num)}")

# === ทดสอบ math module ===
print("\n--- Testing math Module ---")
number_to_sqrt = 16
print(f"Square root of {number_to_sqrt} is {math.sqrt(number_to_sqrt)}")
print(f"Value of pi: {math.pi}")

# === ทดสอบ random module ===
print("\n--- Testing random Module ---")
random_num = random.randint(1, 100)
print(f"Random number between 1 and 100: {random_num}")
```

### Activity 6.1: Refactoring a Previous Lab Assignment
**Format:** Individual coding

**Instructions:** Refactored from Week 2 Lab 2.1 (Number Classifier)
- Original: inline `if/elif` logic
- Refactored: encapsulated into reusable functions `classify_sign()`, `classify_parity()`, `classify_number()`

```python
# ============================================================
# Activity 6.1: Refactoring Week 2 Lab 2.1 (Number Classifier)
# ============================================================

def classify_sign(num):
    """Classifies a number as positive, negative, or zero."""
    if num > 0:
        return "positive"
    elif num < 0:
        return "negative"
    else:
        return "zero"

def classify_parity(num):
    """Classifies a number as even or odd."""
    if num % 2 == 0:
        return "even"
    else:
        return "odd"

def classify_number(num):
    """Returns a formatted string describing the number's sign and parity."""
    sign = classify_sign(num)
    parity = classify_parity(num)
    return f"The number is {sign} and {parity}."


# === Main Program (refactored) ===
num = int(input("Enter an integer: "))
result = classify_number(num)
print(result)
```

### Activity 6.2: Discussion: Benefits of Modular Programming & Standard Library Exploration
**Format:** Class discussion or small group brainstorming

#### Benefits of Modularity

การแบ่งโค้ดออกเป็นฟังก์ชันและโมดูลช่วยได้หลายด้าน:

1. **Code Reusability (การใช้โค้ดซ้ำ):** เขียนฟังก์ชันครั้งเดียว สามารถเรียกใช้ได้หลายที่โดยไม่ต้องเขียนโค้ดเดิมซ้ำ เช่น ฟังก์ชัน `is_prime()` สามารถนำไปใช้ในโปรเจกต์อื่นได้ทันที

2. **Easier Debugging (การดีบั๊กง่ายขึ้น):** เมื่อเกิดบั๊ก สามารถตรวจสอบทีละฟังก์ชันได้ง่าย แทนที่จะต้องไล่โค้ดทั้งหมด

3. **Teamwork (การทำงานเป็นทีม):** แบ่งงานให้สมาชิกแต่ละคนรับผิดชอบโมดูลของตัวเอง เมื่อเสร็จแล้วนำมารวมกัน

4. **Readability (ความอ่านง่าย):** โค้ดที่แบ่งเป็นฟังก์ชันที่ตั้งชื่อดี ทำให้ผู้อ่านเข้าใจได้ง่ายว่าโค้ดแต่ละส่วนทำอะไร

#### Standard Library Exploration

**1. `datetime` module:**
- ใช้จัดการวันที่และเวลา เช่น คำนวณความแตกต่างระหว่างวันที่สองวัน
- ตัวอย่าง: `datetime.datetime.now()` แสดงวันเวลาปัจจุบัน, `datetime.timedelta(days=7)` บวกเวลา 7 วัน

**2. `os` module:**
- ใช้จัดการไฟล์และโฟลเดอร์ในระบบปฏิบัติการ
- ตัวอย่าง: `os.listdir('.')` แสดงรายการไฟล์ในโฟลเดอร์ปัจจุบัน, `os.path.exists('file.txt')` ตรวจสอบว่าไฟล์มีอยู่หรือไม่

**3. `collections` module:**
- มีโครงสร้างข้อมูลพิเศษที่มีประโยชน์มาก
- ตัวอย่าง: `collections.Counter(['a','b','a','c','a'])` นับความถี่ได้ทันทีเป็น `Counter({'a': 3, 'b': 1, 'c': 1})` ซึ่งสะดวกกว่าเขียนลูปนับเอง
